```python
%%html
<!DOCTYPE html>
<html lang="en" class="dark">
<head>
  <meta charset="UTF-8">
  <meta name="viewport" content="width=device-width, initial-scale=1.0">
  <title>Nutrition AI Agent - Harness Engineering Studio</title>
  <script src="https://cdn.tailwindcss.com"></script>
  <link rel="stylesheet" href="https://cdnjs.cloudflare.com/ajax/libs/font-awesome/6.4.0/css/all.min.css">
  <link rel="preconnect" href="https://fonts.googleapis.com">
  <link rel="preconnect" href="https://fonts.gstatic.com" crossorigin>
  <link href="https://fonts.googleapis.com/css2?family=Inter:wght@300;400;500;600;700&family=JetBrains+Mono:wght@400;500;600&display=swap" rel="stylesheet">

  <script>
    tailwind.config = {
      darkMode: 'class',
      theme: {
        extend: {
          fontFamily: {
            sans: ['Inter', 'sans-serif'],
            mono: ['JetBrains Mono', 'monospace'],
          },
          colors: {
            darkbg: '#090d16',
            darkcard: '#111827',
            darkborder: '#1f2937'
          }
        }
      }
    }
  </script>
  <style>
    ::-webkit-scrollbar { width: 6px; height: 6px; }
    ::-webkit-scrollbar-track { background: #0f172a; }
    ::-webkit-scrollbar-thumb { background: #334155; border-radius: 4px; }
    ::-webkit-scrollbar-thumb:hover { background: #475569; }
  </style>
</head>
<body class="bg-darkbg text-slate-100 font-sans antialiased min-h-screen flex flex-col overflow-hidden">

  <!-- TOP NAVBAR -->
  <header class="h-16 border-b border-slate-800 bg-slate-900/90 px-4 lg:px-6 flex items-center justify-between z-30 shrink-0">
    <div class="flex items-center space-x-3">
      <button id="toggle-sidebar-btn" class="lg:hidden text-slate-400 hover:text-white p-2 rounded-lg bg-slate-800">
        <i class="fa-solid fa-bars"></i>
      </button>
      <div class="w-10 h-10 rounded-xl bg-gradient-to-tr from-emerald-600 to-teal-400 flex items-center justify-center shadow-lg shadow-emerald-900/30">
        <i class="fa-solid fa-seedling text-white text-lg"></i>
      </div>
      <div>
        <div class="flex items-center space-x-2">
          <h1 class="font-bold text-base md:text-lg text-white leading-none">NutriHarness AI</h1>
          <span class="text-[10px] uppercase font-mono px-2 py-0.5 rounded-full bg-emerald-500/10 text-emerald-400 border border-emerald-500/30">Harness Eng v2.5</span>
        </div>
        <p class="text-xs text-slate-400">Deterministic Macro Engine + Groq Llama-3.3 Guardrails</p>
      </div>
    </div>

    <div class="flex items-center space-x-3">
      <div class="hidden sm:flex items-center space-x-2 px-3 py-1.5 rounded-lg bg-slate-800/80 border border-slate-700 text-xs">
        <span class="w-2 h-2 rounded-full bg-emerald-400 animate-ping"></span>
        <span class="text-slate-300 font-mono text-[11px]">Guardrail Loop: ACTIVE</span>
      </div>
      <button id="open-settings-btn" class="flex items-center space-x-2 px-3 py-1.5 rounded-lg bg-slate-800 hover:bg-slate-700 text-xs font-medium text-slate-200 border border-slate-700 transition">
        <i class="fa-solid fa-key text-emerald-400"></i>
        <span id="api-key-status" class="hidden md:inline font-mono">Groq API Key</span>
        <i class="fa-solid fa-gear text-slate-400"></i>
      </button>
    </div>
  </header>

  <!-- MAIN WRAPPER -->
  <div class="flex-1 flex overflow-hidden relative">

    <!-- LEFT SIDEBAR -->
    <aside id="sidebar" class="w-full lg:w-96 bg-slate-900/95 border-r border-slate-800 flex flex-col z-20 absolute lg:relative inset-y-0 left-0 transform -translate-x-full lg:translate-x-0 transition-transform duration-300 ease-in-out overflow-y-auto">
      <div class="p-4 border-b border-slate-800 flex items-center justify-between">
        <div class="flex items-center space-x-2">
          <i class="fa-solid fa-sliders text-emerald-400"></i>
          <h2 class="font-semibold text-sm uppercase tracking-wider text-slate-200">User Profile & Engine</h2>
        </div>
        <button id="close-sidebar-btn" class="lg:hidden text-slate-400 hover:text-white p-1">
          <i class="fa-solid fa-xmark"></i>
        </button>
      </div>

      <div class="p-4 space-y-4">
        <div class="grid grid-cols-2 gap-3">
          <div>
            <label class="block text-xs font-medium text-slate-400 mb-1">Age (years)</label>
            <input type="number" id="user-age" value="26" min="15" max="95" class="w-full bg-slate-950 border border-slate-800 rounded-lg px-3 py-1.5 text-sm text-white focus:outline-none focus:border-emerald-500">
          </div>
          <div>
            <label class="block text-xs font-medium text-slate-400 mb-1">Gender</label>
            <select id="user-gender" class="w-full bg-slate-950 border border-slate-800 rounded-lg px-3 py-1.5 text-sm text-white focus:outline-none focus:border-emerald-500">
              <option value="male" selected>Male</option>
              <option value="female">Female</option>
            </select>
          </div>
        </div>

        <div class="grid grid-cols-2 gap-3">
          <div>
            <label class="block text-xs font-medium text-slate-400 mb-1">Weight (kg)</label>
            <input type="number" id="user-weight" value="78" step="0.5" class="w-full bg-slate-950 border border-slate-800 rounded-lg px-3 py-1.5 text-sm text-white focus:outline-none focus:border-emerald-500">
          </div>
          <div>
            <label class="block text-xs font-medium text-slate-400 mb-1">Height (cm)</label>
            <input type="number" id="user-height" value="175" class="w-full bg-slate-950 border border-slate-800 rounded-lg px-3 py-1.5 text-sm text-white focus:outline-none focus:border-emerald-500">
          </div>
        </div>

        <div>
          <label class="block text-xs font-medium text-slate-400 mb-1">Activity Level</label>
          <select id="user-activity" class="w-full bg-slate-950 border border-slate-800 rounded-lg px-3 py-1.5 text-sm text-white focus:outline-none focus:border-emerald-500">
            <option value="sedentary">Sedentary (Office desk job)</option>
            <option value="light">Lightly Active (1-3 days exercise)</option>
            <option value="moderate" selected>Moderately Active (3-5 days workout)</option>
            <option value="active">Very Active (6-7 days intense sports)</option>
            <option value="very_active">Extremely Active (Physical labor / Athlete)</option>
          </select>
        </div>

        <div>
          <label class="block text-xs font-medium text-slate-400 mb-1">Fitness Goal</label>
          <select id="user-goal" class="w-full bg-slate-950 border border-slate-800 rounded-lg px-3 py-1.5 text-sm text-white focus:outline-none focus:border-emerald-500">
            <option value="fat_loss" selected>Fat Loss (-450 kcal deficit)</option>
            <option value="maintenance">Maintenance (Maintain TDEE)</option>
            <option value="muscle_gain">Muscle Gain (+300 kcal surplus)</option>
          </select>
        </div>

        <div>
          <label class="block text-xs font-medium text-slate-400 mb-1">Dietary Preference</label>
          <input type="text" id="user-diet" value="South Indian Non-Veg (Eggs, Chicken, Rice, Idli)" class="w-full bg-slate-950 border border-slate-800 rounded-lg px-3 py-1.5 text-sm text-white focus:outline-none focus:border-emerald-500">
        </div>

        <div>
          <label class="block text-xs font-medium text-slate-400 mb-1">Allergies / Restrictions</label>
          <input type="text" id="user-allergies" value="Peanuts, Shellfish" class="w-full bg-slate-950 border border-slate-800 rounded-lg px-3 py-1.5 text-sm text-white focus:outline-none focus:border-emerald-500">
        </div>
      </div>

      <!-- DETERMINISTIC MACRO OUTPUT -->
      <div class="p-4 border-t border-slate-800 bg-slate-950/60 mt-auto">
        <div class="flex items-center justify-between mb-3">
          <span class="text-xs font-bold uppercase tracking-wider text-emerald-400 flex items-center gap-1.5">
            <i class="fa-solid fa-calculator"></i> Harness Math Engine
          </span>
          <span class="text-[10px] font-mono text-slate-400 bg-slate-800 px-2 py-0.5 rounded">Mifflin-St Jeor</span>
        </div>

        <div class="grid grid-cols-2 gap-2 text-xs mb-3">
          <div class="bg-slate-900 p-2.5 rounded-lg border border-slate-800">
            <div class="text-slate-400 text-[11px]">BMR (Basal)</div>
            <div id="stat-bmr" class="font-mono font-bold text-slate-200 text-sm">1,723 kcal</div>
          </div>
          <div class="bg-slate-900 p-2.5 rounded-lg border border-slate-800">
            <div class="text-slate-400 text-[11px]">TDEE (Total)</div>
            <div id="stat-tdee" class="font-mono font-bold text-slate-200 text-sm">2,670 kcal</div>
          </div>
        </div>

        <div class="bg-gradient-to-r from-emerald-950/50 to-slate-900 p-3 rounded-xl border border-emerald-500/20 mb-2">
          <div class="flex justify-between items-baseline mb-2">
            <span class="text-xs text-slate-300 font-medium">Daily Target</span>
            <span id="stat-target-cal" class="font-mono text-base font-extrabold text-emerald-400">2,220 kcal</span>
          </div>
          <div class="space-y-1.5">
            <div>
              <div class="flex justify-between text-[11px] mb-0.5">
                <span class="text-slate-400">Protein (1.8g/kg)</span>
                <span id="stat-protein" class="font-mono text-slate-200">140.4g</span>
              </div>
              <div class="w-full bg-slate-800 h-1.5 rounded-full overflow-hidden">
                <div id="bar-protein" class="bg-emerald-500 h-full" style="width: 25%"></div>
              </div>
            </div>
            <div>
              <div class="flex justify-between text-[11px] mb-0.5">
                <span class="text-slate-400">Fats (25% cals)</span>
                <span id="stat-fats" class="font-mono text-slate-200">61.7g</span>
              </div>
              <div class="w-full bg-slate-800 h-1.5 rounded-full overflow-hidden">
                <div id="bar-fats" class="bg-amber-400 h-full" style="width: 25%"></div>
              </div>
            </div>
            <div>
              <div class="flex justify-between text-[11px] mb-0.5">
                <span class="text-slate-400">Carbs (Remainder)</span>
                <span id="stat-carbs" class="font-mono text-slate-200">276.0g</span>
              </div>
              <div class="w-full bg-slate-800 h-1.5 rounded-full overflow-hidden">
                <div id="bar-carbs" class="bg-blue-400 h-full" style="width: 50%"></div>
              </div>
            </div>
          </div>
        </div>
      </div>
    </aside>

    <div id="sidebar-backdrop" class="hidden fixed inset-0 bg-black/60 z-10 lg:hidden"></div>

    <!-- MAIN CHAT AREA -->
    <main class="flex-1 flex flex-col bg-darkbg h-full overflow-hidden relative">
      <div id="chat-container" class="flex-1 overflow-y-auto p-4 md:p-6 space-y-6">
        <div id="welcome-card" class="bg-gradient-to-br from-slate-900 via-slate-900/90 to-emerald-950/40 border border-slate-800 rounded-2xl p-5 shadow-xl max-w-3xl mx-auto">
          <div class="flex items-start space-x-4">
            <div class="w-12 h-12 rounded-2xl bg-emerald-500/20 border border-emerald-500/30 flex items-center justify-center text-emerald-400 shrink-0">
              <i class="fa-solid fa-robot text-xl"></i>
            </div>
            <div>
              <h2 class="text-lg font-bold text-white mb-1">NutriHarness AI Agent</h2>
              <p class="text-xs md:text-sm text-slate-300 leading-relaxed mb-3">
                Runs <strong>L1 Input Guardrail</strong>, <strong>L2 Deterministic Macro Math</strong>, <strong>L3 Groq Llama-3.3 JSON Generation</strong>, and <strong>L4 Calorie Verifier + Retry Loop</strong>.
              </p>
              <div class="flex flex-wrap gap-2">
                <button class="chip-btn text-xs bg-slate-950 hover:bg-slate-800 text-slate-300 border border-slate-800 px-3 py-1.5 rounded-xl transition">
                  🍱 South Indian fat loss meal plan
                </button>
                <button class="chip-btn text-xs bg-slate-950 hover:bg-slate-800 text-slate-300 border border-slate-800 px-3 py-1.5 rounded-xl transition">
                  💪 High protein muscle gain meals
                </button>
              </div>
            </div>
          </div>
        </div>

        <div id="messages-list" class="max-w-3xl mx-auto space-y-6"></div>
      </div>

      <!-- CHAT INPUT -->
      <div class="p-3 md:p-4 border-t border-slate-800 bg-slate-900/90 shrink-0">
        <div class="max-w-3xl mx-auto">
          <form id="chat-form" class="flex items-end bg-slate-950 border border-slate-800 focus-within:border-emerald-500/70 rounded-2xl p-2 transition shadow-lg">
            <textarea id="user-input" rows="2" placeholder="Ask NutriHarness AI... (e.g. Enakku fat loss diet plan kudungada)" class="w-full bg-transparent text-slate-100 placeholder-slate-500 text-sm px-3 py-1 focus:outline-none resize-none"></textarea>
            <div class="flex items-center space-x-2 shrink-0 pb-1 pr-1">
              <button type="button" id="clear-chat-btn" title="Clear Chat" class="text-slate-500 hover:text-slate-300 p-2 rounded-lg transition">
                <i class="fa-solid fa-trash-can"></i>
              </button>
              <button type="submit" id="send-btn" class="bg-emerald-600 hover:bg-emerald-500 text-white font-medium p-2.5 rounded-xl transition flex items-center justify-center">
                <i class="fa-solid fa-paper-plane text-sm"></i>
              </button>
            </div>
          </form>
        </div>
      </div>
    </main>
  </div>

  <!-- SETTINGS MODAL -->
  <div id="settings-modal" class="hidden fixed inset-0 bg-black/70 backdrop-blur-sm z-50 flex items-center justify-center p-4">
    <div class="bg-slate-900 border border-slate-800 rounded-2xl max-w-md w-full p-6 shadow-2xl relative">
      <button id="close-settings-btn" class="absolute top-4 right-4 text-slate-400 hover:text-white">
        <i class="fa-solid fa-xmark text-lg"></i>
      </button>
      <h3 class="font-bold text-slate-100 text-base mb-3">Groq API Key</h3>
      <input type="password" id="groq-key-input" placeholder="gsk_..." class="w-full bg-slate-950 border border-slate-800 rounded-lg px-3 py-2 text-xs font-mono text-white focus:outline-none focus:border-emerald-500 mb-3">
      <div class="flex justify-end space-x-2">
        <button id="save-key-btn" class="px-4 py-1.5 text-xs bg-emerald-600 hover:bg-emerald-500 text-white font-medium rounded-lg transition">Save Key</button>
      </div>
    </div>
  </div>

  <script>
    const GROQ_MODEL = "llama-3.3-70b-versatile";
    const GROQ_ENDPOINT = "https://api.groq.com/openai/v1/chat/completions";

    let groqApiKey = localStorage.getItem("GROQ_API_KEY") || "";

    const sidebar = document.getElementById("sidebar");
    const sidebarBackdrop = document.getElementById("sidebar-backdrop");
    const toggleSidebarBtn = document.getElementById("toggle-sidebar-btn");
    const closeSidebarBtn = document.getElementById("close-sidebar-btn");
    const settingsModal = document.getElementById("settings-modal");
    const openSettingsBtn = document.getElementById("open-settings-btn");
    const closeSettingsBtn = document.getElementById("close-settings-btn");
    const saveKeyBtn = document.getElementById("save-key-btn");
    const groqKeyInput = document.getElementById("groq-key-input");
    const chatForm = document.getElementById("chat-form");
    const userInput = document.getElementById("user-input");
    const messagesList = document.getElementById("messages-list");
    const chatContainer = document.getElementById("chat-container");
    const clearChatBtn = document.getElementById("clear-chat-btn");
    const welcomeCard = document.getElementById("welcome-card");

    const inputAge = document.getElementById("user-age");
    const inputGender = document.getElementById("user-gender");
    const inputWeight = document.getElementById("user-weight");
    const inputHeight = document.getElementById("user-height");
    const inputActivity = document.getElementById("user-activity");
    const inputGoal = document.getElementById("user-goal");
    const inputDiet = document.getElementById("user-diet");
    const inputAllergies = document.getElementById("user-allergies");

    function updateApiKeyBadge() {
      const badge = document.getElementById("api-key-status");
      if (groqApiKey) {
        badge.innerText = "API Key Set";
        badge.classList.add("text-emerald-400");
      } else {
        badge.innerText = "Set API Key";
        badge.classList.remove("text-emerald-400");
      }
    }
    updateApiKeyBadge();

    toggleSidebarBtn.addEventListener("click", () => {
      sidebar.classList.remove("-translate-x-full");
      sidebarBackdrop.classList.remove("hidden");
    });
    closeSidebarBtn.addEventListener("click", () => {
      sidebar.classList.add("-translate-x-full");
      sidebarBackdrop.classList.add("hidden");
    });
    sidebarBackdrop.addEventListener("click", () => {
      sidebar.classList.add("-translate-x-full");
      sidebarBackdrop.classList.add("hidden");
    });

    openSettingsBtn.addEventListener("click", () => {
      groqKeyInput.value = groqApiKey;
      settingsModal.classList.remove("hidden");
    });
    closeSettingsBtn.addEventListener("click", () => settingsModal.classList.add("hidden"));
    saveKeyBtn.addEventListener("click", () => {
      groqApiKey = groqKeyInput.value.trim();
      localStorage.setItem("GROQ_API_KEY", groqApiKey);
      updateApiKeyBadge();
      settingsModal.classList.add("hidden");
    });

    function calculateMacros() {
      const age = parseInt(inputAge.value) || 25;
      const gender = inputGender.value;
      const weight = parseFloat(inputWeight.value) || 70;
      const height = parseFloat(inputHeight.value) || 170;
      const activity = inputActivity.value;
      const goal = inputGoal.value;

      let bmr = gender === "male"
        ? (10 * weight) + (6.25 * height) - (5 * age) + 5
        : (10 * weight) + (6.25 * height) - (5 * age) - 161;

      const mults = { sedentary: 1.2, light: 1.375, moderate: 1.55, active: 1.725, very_active: 1.9 };
      const tdee = bmr * (mults[activity] || 1.2);

      let targetCals = tdee;
      if (goal === "fat_loss") targetCals = tdee - 450;
      else if (goal === "muscle_gain") targetCals = tdee + 300;
      targetCals = Math.round(targetCals);

      const proteinG = Math.round((weight * 1.8) * 10) / 10;
      const proteinCals = proteinG * 4;
      const fatCals = targetCals * 0.25;
      const fatsG = Math.round((fatCals / 9) * 10) / 10;
      const carbCals = Math.max(targetCals - proteinCals - fatCals, 200);
      const carbsG = Math.round((carbCals / 4) * 10) / 10;

      document.getElementById("stat-bmr").innerText = `${Math.round(bmr).toLocaleString()} kcal`;
      document.getElementById("stat-tdee").innerText = `${Math.round(tdee).toLocaleString()} kcal`;
      document.getElementById("stat-target-cal").innerText = `${targetCals.toLocaleString()} kcal`;
      document.getElementById("stat-protein").innerText = `${proteinG}g`;
      document.getElementById("stat-fats").innerText = `${fatsG}g`;
      document.getElementById("stat-carbs").innerText = `${carbsG}g`;

      const total = proteinCals + fatCals + carbCals || 1;
      document.getElementById("bar-protein").style.width = `${Math.round((proteinCals / total) * 100)}%`;
      document.getElementById("bar-fats").style.width = `${Math.round((fatCals / total) * 100)}%`;
      document.getElementById("bar-carbs").style.width = `${Math.round((carbCals / total) * 100)}%`;

      return {
        age, gender, weight, height, activity, goal,
        bmr: Math.round(bmr), tdee: Math.round(tdee),
        targetCalories: targetCals, proteinG, fatsG, carbsG,
        dietPreference: inputDiet.value, allergies: inputAllergies.value
      };
    }

    [inputAge, inputGender, inputWeight, inputHeight, inputActivity, inputGoal, inputDiet, inputAllergies].forEach(el => {
      el.addEventListener("input", calculateMacros);
      el.addEventListener("change", calculateMacros);
    });
    calculateMacros();

    document.querySelectorAll(".chip-btn").forEach(btn => {
      btn.addEventListener("click", () => {
        userInput.value = btn.innerText.replace(/^[^\w]+/, "").trim();
        userInput.focus();
      });
    });

    clearChatBtn.addEventListener("click", () => {
      messagesList.innerHTML = "";
      welcomeCard.classList.remove("hidden");
    });

    userInput.addEventListener("keydown", (e) => {
      if (e.key === "Enter" && !e.shiftKey) {
        e.preventDefault();
        chatForm.dispatchEvent(new Event("submit"));
      }
    });

    async function runInputGuardrail(query, profile) {
      const lower = query.toLowerCase();
      const blockedWords = ["anorexia", "bulimia", "starve", "0 calories", "vomit", "suicide"];
      for (const w of blockedWords) {
        if (lower.includes(w)) {
          return { isSafe: false, reason: `Blocked by local guardrail: '${w}'` };
        }
      }

      const res = await fetch(GROQ_ENDPOINT, {
        method: "POST",
        headers: {
          "Authorization": `Bearer ${groqApiKey}`,
          "Content-Type": "application/json"
        },
        body: JSON.stringify({
          model: GROQ_MODEL,
          messages: [
            {
              role: "system",
              content: "You are a nutrition safety guardrail. Return JSON: {\"is_safe\": boolean, \"reason\": string}. Block eating disorders, extreme starvation, or medical emergencies."
            },
            {
              role: "user",
              content: `Query: ${query}\nProfile: ${JSON.stringify(profile)}`
            }
          ],
          response_format: { type: "json_object" },
          temperature: 0
        })
      });

      if (!res.ok) {
        const err = await res.text();
        throw new Error(`Groq API error (${res.status}): ${err}`);
      }
      const data = await res.json();
      const parsed = JSON.parse(data.choices[0].message.content);
      return { isSafe: parsed.is_safe !== false, reason: parsed.reason || "Safe" };
    }

    async function generatePlan(query, metrics, feedback = "") {
      const prompt = `
Create a structured daily nutrition plan in strict JSON format:
{
  "summary": "short helpful note",
  "meals": [
    {"meal_time": "Breakfast", "items": ["item 1", "item 2"], "calories": 500, "protein_g": 35, "carbs_g": 55, "fats_g": 14}
  ],
  "hydration": "water intake advice"
}

User Query: ${query}
Diet Preference: ${metrics.dietPreference}
Allergies to AVOID: ${metrics.allergies}
Mandatory Targets:
- Target Calories: ${metrics.targetCalories} kcal (sum of meal calories must be within ±150 kcal of ${metrics.targetCalories})
- Protein: ${metrics.proteinG}g
- Carbs: ${metrics.carbsG}g
- Fats: ${metrics.fatsG}g
${feedback ? `Fix previous verifier issue: ${feedback}` : ""}
`;

      const res = await fetch(GROQ_ENDPOINT, {
        method: "POST",
        headers: {
          "Authorization": `Bearer ${groqApiKey}`,
          "Content-Type": "application/json"
        },
        body: JSON.stringify({
          model: GROQ_MODEL,
          messages: [
            { role: "system", content: "You are a precise clinical nutritionist. Respond ONLY with valid JSON." },
            { role: "user", content: prompt }
          ],
          response_format: { type: "json_object" },
          temperature: 0.3
        })
      });

      if (!res.ok) {
        const err = await res.text();
        throw new Error(`Groq API error (${res.status}): ${err}`);
      }
      const data = await res.json();
      return JSON.parse(data.choices[0].message.content);
    }

    function verifyPlan(plan, metrics) {
      const minSafe = metrics.gender === "female" ? 1200 : 1500;
      if (metrics.targetCalories < minSafe) {
        return { ok: false, reason: `Target calories below safe minimum (${minSafe} kcal).` };
      }
      const meals = Array.isArray(plan.meals) ? plan.meals : [];
      const sumCals = meals.reduce((acc, m) => acc + (Number(m.calories) || 0), 0);
      if (Math.abs(sumCals - metrics.targetCalories) > 150) {
        return { ok: false, reason: `Meal sum (${sumCals} kcal) differs from target (${metrics.targetCalories} kcal) by more than 150 kcal.` };
      }
      return { ok: true, reason: `Verified (${sumCals} kcal total).` };
    }

    function appendUserBubble(text) {
      welcomeCard.classList.add("hidden");
      const div = document.createElement("div");
      div.className = "flex justify-end";
      div.innerHTML = `<div class="bg-emerald-600 text-white px-4 py-3 rounded-2xl rounded-br-sm max-w-xl text-sm shadow">${escapeHtml(text)}</div>`;
      messagesList.appendChild(div);
      chatContainer.scrollTop = chatContainer.scrollHeight;
    }

    function appendBotBubble(htmlContent) {
      const div = document.createElement("div");
      div.className = "flex justify-start";
      div.innerHTML = `<div class="bg-slate-900 border border-slate-800 text-slate-100 p-4 rounded-2xl rounded-bl-sm max-w-2xl w-full text-sm shadow-lg">${htmlContent}</div>`;
      messagesList.appendChild(div);
      chatContainer.scrollTop = chatContainer.scrollHeight;
      return div;
    }

    function escapeHtml(str) {
      return String(str).replace(/[&<>"']/g, s => ({ "&": "&", "<": "<", ">": ">", '"': """, "'": "'" }[s]));
    }

    chatForm.addEventListener("submit", async (e) => {
      e.preventDefault();
      const query = userInput.value.trim();
      if (!query) return;

      if (!groqApiKey) {
        settingsModal.classList.remove("hidden");
        return;
      }

      userInput.value = "";
      appendUserBubble(query);
      const metrics = calculateMacros();
      const loader = appendBotBubble(`<div class="text-slate-400 text-xs font-mono"><i class="fa-solid fa-spinner fa-spin mr-2 text-emerald-400"></i>Running Harness Pipeline (Guardrail -> Math -> Groq -> Verifier)...</div>`);

      try {
        const guard = await runInputGuardrail(query, metrics);
        if (!guard.isSafe) {
          loader.remove();
          appendBotBubble(`<div class="text-rose-400 font-semibold mb-1"><i class="fa-solid fa-ban mr-1"></i> Blocked by Input Guardrail</div><p class="text-xs text-slate-300">${escapeHtml(guard.reason)}</p>`);
          return;
        }

        let feedback = "";
        let plan = null;
        let check = { ok: false, reason: "" };

        for (let attempt = 1; attempt <= 2; attempt++) {
          plan = await generatePlan(query, metrics, feedback);
          check = verifyPlan(plan, metrics);
          if (check.ok) break;
          feedback = check.reason;
        }

        loader.remove();
        const mealsHtml = (plan.meals || []).map(m => `
          <div class="bg-slate-950/80 border border-slate-800 rounded-xl p-3">
            <div class="flex justify-between items-center mb-1">
              <span class="font-semibold text-emerald-400 text-xs uppercase">${escapeHtml(m.meal_time || "Meal")}</span>
              <span class="font-mono text-xs text-slate-300">${Number(m.calories) || 0} kcal</span>
            </div>
            <div class="text-xs text-slate-200 mb-1.5">${escapeHtml((m.items || []).join(", "))}</div>
            <div class="text-[11px] font-mono text-slate-400">P: ${m.protein_g ?? 0}g · C: ${m.carbs_g ?? 0}g · F: ${m.fats_g ?? 0}g</div>
          </div>
        `).join("");

        appendBotBubble(`
          <div class="flex items-center justify-between mb-2">
            <span class="text-xs font-mono text-emerald-400"><i class="fa-solid fa-circle-check mr-1"></i>${escapeHtml(check.reason)}</span>
            <span class="text-[11px] font-mono text-slate-400">Target: ${metrics.targetCalories} kcal</span>
          </div>
          <p class="text-xs text-slate-300 mb-3">${escapeHtml(plan.summary || "")}</p>
          <div class="space-y-2 mb-3">${mealsHtml}</div>
          <div class="text-xs text-slate-400 border-t border-slate-800 pt-2">💧 ${escapeHtml(plan.hydration || "Drink 2.5–3L water daily.")}</div>
        `);
      } catch (err) {
        loader.remove();
        appendBotBubble(`<div class="text-rose-400 font-semibold mb-1"><i class="fa-solid fa-triangle-exclamation mr-1"></i> Execution Error</div><pre class="text-xs text-slate-300 whitespace-pre-wrap font-mono">${escapeHtml(err.message)}</pre>`);
      }
    });
  </script>
</body>
</html>
```
